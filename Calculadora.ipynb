```python
def calculator():
    print("Welcome to your virtual calculator, my name is: 'Calcus'")
    print("What operation would you like for me to do?")
    print("Please input two numbers and the type of operation you are trying to accomplish")

    def add(x, y):
        return x + y

    def subtract(x, y):
        return x - y

    def multiply(x, y):
        return x * y

    def divide(x, y):
        if y != 0:
            return x / y
        else:
            return "Calcus states: Error: Sorry, division by zero is not allowed"

    def to_hex(x):
        return hex(int(x))

    def to_binary(x):
        return bin(int(x))

    def to_octal(x):
        return oct(int(x))

    while True:
        print("Select operation.")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Convert to Hexadecimal")
        print("6. Convert to Binary")
        print("7. Convert to Octal")
        print("8. Exit")

        choice = input("Calcus states: Enter choice (1/2/3/4/5/6/7/8): ")

        if choice == '8':
            print("Calcus states: Exiting the calculator. Thank you, User! Goodbye!")
            break

        if choice not in ['1', '2', '3', '4', '5', '6', '7', '8']:
            print("Invalid input. Please enter a valid operation (1/2/3/4/5/6/7/8).")
            continue

        try:
            if choice in ['5', '6', '7']:
                num = float(input("Calcus states: Enter your number: "))
                if choice == '5':
                    result = f"{num} in hexadecimal is {to_hex(num)}"
                elif choice == '6':
                    result = f"{num} in binary is {to_binary(num)}"
                elif choice == '7':
                    result = f"{num} in octal is {to_octal(num)}"
            else:
                num1 = float(input("Calcus states: Enter your first number: "))
                num2 = float(input("Calcus states: Enter your second number: "))

                if choice == '1':
                    result = f"{num1} + {num2} = {add(num1, num2)}"
                elif choice == '2':
                    result = f"{num1} - {num2} = {subtract(num1, num2)}"
                elif choice == '3':
                    result = f"{num1} * {num2} = {multiply(num1, num2)}"
                elif choice == '4':
                    result = divide(num1, num2)
                    if result == "Error: Division by zero is not allowed":
                        print(result)
                        continue
                    result = f"{num1} / {num2} = {result}"

            print(result)
        except ValueError:
            print("Invalid input. Please enter numeric values for the numbers.")

calculator()
```
