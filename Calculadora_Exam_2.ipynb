```python
def calculator():
    print("Welcome to your virtual calculator, my name is: 'Calcus'")
    print("What operation would you like me to perform?")
    print("Please enter two numbers and the type of operation you want to accomplish")

    def add(x, y):
        return x + y

    def subtract(x, y):
        return x - y

    def multiply(x, y):
        return x * y

    def divide(x, y):
        if y != 0:
            return x / y
        else:
            return "Error: Division by zero is not allowed"

    def power(x, y):
        return x ** y

    while True:
        print("Select operation.")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Power")
        print("6. Exit")

        choice = input("Calcus states: Enter choice (1/2/3/4/5/6): ")

        if choice == '6':
            print("Exiting the calculator. Goodbye!")
            break

        if choice not in ['1', '2', '3', '4', '5', '6']:
            print("Invalid input. Please enter a valid operation (1/2/3/4/5/6).")
            continue

        lista = input("Enter a list of 5 numbers separated by commas: ")
        set_ = input("Enter a set of 5 numbers separated by commas: ")
        diccionario = input("Enter a dictionary of 5 numbers in key:value format separated by commas: ")

        try:
            lista = [float(x) for x in lista.split(",")]
            set_ = {float(x) for x in set_.split(",")}
            diccionario = {x.split(":")[0]: float(x.split(":")[1]) for x in diccionario.split(",")}
        except ValueError:
            print("Invalid input")
            continue

        if len(lista) != 5 or len(set_) != 5 or len(diccionario) != 5:
            print("Invalid input size")
            continue

        if choice == '1':
            resultado = sum(lista) + sum(set_) + sum(diccionario.values())
        elif choice == '2':
            resultado = lista[0] - sum(set_) - sum(diccionario.values())
        elif choice == '3':
            resultado = 1
            for x in lista:
                resultado *= x
            for x in set_:
                resultado *= x
            for x in diccionario.values():
                resultado *= x
        elif choice == '4':
            resultado = lista[0]
            for x in set_:
                resultado /= x
            for x in diccionario.values():
                resultado /= x
            if resultado == "Error: Division by zero is not allowed":
                print(resultado)
                continue
        elif choice == '5':
            resultado = lista[0]
            for x in set_:
                resultado **= x
            for x in diccionario.values():
                resultado **= x

        print("Input values:")
        print(f"List: {lista}")
        print(f"Set: {set_}")
        print(f"Dictionary: {diccionario}")
        print(f"Operation: {choice}")
        print(f"Result: {resultado}")

calculator()
```
