# Variables

## Que es una variable?
- Son un espacio temporal para guardar valor con un nombre especifico
- No hay necesicidad de declarar el tipo de dato explicitamente
- Python es un lenguaje tipado dinamicamente, lo que quiere decir que el tipo de dato se define mientras corre el programa. 
- Las variables de Python funcionan diferentes que otros lenjuages de programacion. 
- No necesitamos crear variable, solo asignamos un nombre a un valor usando el operador de asignacion.

### Tipos de variables
- Variables globales: aquellas que se utilizaran para todo el programa
- Variables locales: aquellas que se van a utilizar solo en un bloque especifico

# Variables Globales
- Declaramos variables globales afuera de funciones 
- El valor de la variable puede ser constante para todo el programa
- Cualquier funcion podra acceder a estas variables
- Podemos decir que las variables Globales tienen un alcance global
- Normalmente son mas usadas para constantes
- Las Variables globales mantienen sus valores atravez de la ejecucion complete de tu programa
- Una Variable Global puede ser accesada por cualquier funcion

```python
# Variable Global es cuando esta fuera de una funcion. 
my_global_variable = 50

def my_function_1():
    pass

def my_function_2():
    pass
```

```python
# Variables locales
def my_function_1():
    my_local_variable = 5

def my_function_2():
    my_local_variable = 35
```

# Palabra reservada "global"

```python
def my_function_1():
    my_variable=10 #local
    print("my_function_1 - El valor de my_variable: ", my_variable)

def my_function_2():
    global my_variable #Global
    my_variable=20 
    print("my_function_2 - El valor de my_variable: ", my_variable)
        
def my_function_3():
    print("my_function_3 - El valor de my_variable: ", my_variable)
```
