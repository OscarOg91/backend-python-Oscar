# Operadores y Tipos de Operadores

Que son los operadores?
- Son un conjunto de simbolos para un lenguaje de programacion
- Pueden decirle al compilador o interprete que ejecute ciertas acciones para producir un resultaod final
- Pueden ser usado para aplicar operaciones en variables o valores
- Pueden ser clasifacados segun su tipo (aritmeticos, relacionales, de asignacion, logicos, afiliacion y de identidad)

Tipos de Operadores:
- Aritmeticos
- Relacionales
- De asignacion
- Logicos
- De afiliacion
- De identidad

### Operadores aritmeticos
- "+": Suma
- "-": Resta
- "*": Multiplicacion
- "/": Division
- "//": Division con redondeo hacia abajo
- "**": Exponente

```python
# "+": Suma
num_1= 500
num_2=300
print("El resultado de num_1 + num_2 es:", num_1+num_2)
```

```python
# "+": Suma
num_1=200
num_2=900
print("El resultado de num_1 + num_2 es:", num_1+num_2)
```

```python
# "-": Resta
num_1=500
num_2=300
print("El resultado de num_1 - num_2 es:", num_1-num_2)
```

```python
# "*": Mult
num_1=500
num_2=300
print("El resultado de num_1 * num_2 es:", num_1*num_2)
```

```python
# "/": Division
num_1=500
num_2=300
print("El resultado de num_1 * num_2 es:", num_1/num_2)
```

```python
# "//": Division con redondeo hacia abajo, la parte entera del result
num_1=500
num_2=300
print("El resultado de num_1 * num_2 es:", num_1//num_2)
```

```python
# "**": Exponente (Elevar un exponente)
num_1=2
num_2=5
print("El resultado de num_1 ** num_2 es:", num_1**num_2) # 2*2*2*2*2
```

### Operadores Relacionales

Que son los operadores Relacionales?
- Son tambien conocidos como Operadores de comparacion
- Regresan un valor booleano (True or False)
- En Python, son usados para comparar o relacionar los valores
- Verdadero/True es el resultado cuando la concidion es exitosa
- Falso/false, es el resultado cuando la condicion no se cumple. 
- Estos operadores se encargan de las operaciones de comparacion

Tipos de operadores relacionales:
- "=="": Igual a
- "!=": Diferenta a
- ">": Mayor que
- ">=": Mayor o igual que
- "<": Menor que
- "<=": Menor o igual que

## Operadores Logicos

Que son lo operadores logicos?
- Son usados principalmente para declaraciones condicionales
- Son indicadores estructurales
- Tambien indicadores estructurales
- Hayu 3 tipos de Operadores logicos:
    - AND (Y)- Regresa True/Verdadero si ambos operaciones son verdaderas
    - OR (O) - Rgresa True/Verdadero si por lo menos una de las operaciones son verdaderas
    - NOT (Lo Opuesto) - Regresa True/Verdadero si la operacion es False
    

```python
# AND (y)
# Todas las condiciones deben de ser verdaderas
```

### Operadores de afiliaciones

Que son lo operadores de Afilicion?
- Son usados para validar los elementos en listas, cadenas y tuplas
- Validan si una variable esta presente en una sequencia
- Regresan un valor booleano con el resultado que puede ser true or false
- Regresan True si el valor esta presente en la sequiencia
- Regresan False si el valor no esta presente en la sequiencia. 
- Hay 2 operadores de afiliacion:
    - IN - El valor esta en la sequiencia
    - NOT IN - El valor no esta en la sequencia


```python
# IN: El valor esta presente en la sequencia?
```

```python
my_string="Backend Python"
print("'a' in my_string: ",'a' in my_string)
```

```python
my_list=[1, 154, 321, 22, 11, 5, 11, 22, 5, 1]
# Debe ser el valor de 5, no 154
print("5 in my_list?: ", 5 in my_list)
```

```python
my_list
```

```python
my_set= {1, 154, 321, 22, 11, 5, 11, 22, 5, 1}
# Guarda valores unicos
print("5 in my_set",5 in my_set)
```

```python
my_set
```

```python
# NOT IN: El valor NO esta presente en la sequencia?
```

```python
my_string="Backend Python"
# False bc a is inside my_string
print("'a' not my_string: ",'a' not in my_string)
```

```python
my_string="Backend Python"
# True bc a is not inside my_string
print("'i' not my_string: ",'i' not in my_string)
```

```python
my_list=[1, 154, 321, 22, 11]
# Debe ser el valor de 5, no 154
print("5 not in my_list?: ", 5 not in my_list)
```

### Operadores de identidad

Que son los operadores de identidad?
- Son usados para comparar la ubicacion en memoria de dos objectos
- Se utilizan para validar si los valores comparten el mismo espacio en memoria o no
- Regresa True si lo comparte, y False si no lo comparten
- Hay dos operadores de identidad
    - IS
    - IS NOT

```python
num_1=11
num_2=55
print("num_1 is num_2", num_1 is num_2)
```

```python
num_1=77
num_2=77
print("num_1 is num_2", num_1 is num_2)
```

```python
# IS NOT
```

```python
num_1=11
num_2=55
print("num_1 is not num_2", num_1 is not num_2)
```

```python
num_1=77
num_2=77
print("num_1 is not num_2", num_1 is not num_2)
```

# Funciones basicas

### trunc()

```python
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("num_1: ", num_1)
print("num_2: ", num_2)
print("num_3: ", num_3)
print("num_4: ", num_4)
```

```python
from math import trunc
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("trunc(num_1): ", trunc(num_1))
print("trunc(num_2): ", trunc(num_2))
print("trunc(num_3): ", trunc(num_3))
print("trunc(num_4): ", trunc(num_4))
```

### round(), ceil() y float()

```python
# round
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("round(num_1): ", round(num_1))
print("round(num_2): ", round(num_2))
print("round(num_3): ", round(num_3))
print("round(num_4): ", round(num_4))
```

```python
# ceil - redondeo hacia arriba
from math import ceil
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("ceil(num_1): ", ceil(num_1))
print("ceil(num_2): ", ceil(num_2))
print("ceil(num_3): ", ceil(num_3))
print("ceil(num_4): ", ceil(num_4))
```

```python
# floor - redondeo hacia abajo
from math import floor
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("floor(num_1): ", floor(num_1))
print("floor(num_2): ", floor(num_2))
print("floor(num_3): ", floor(num_3))
print("floor(num_4): ", floor(num_4))
```

### Funcion %

```python
from math import floor
num_1=64.99
num_2=64.5
num_3=64.54
num_4=64.0000000000001
print("'%.1f' % num_1: ", '%.1f' % num_1)
print("'%.3f' % num_3: ", '%.3f' % num_2)
print("'%.4f' % num_4: ", '%.4f' % num_3)
print("'%.5f' % num_5: ", '%.5f' % num_4)
```

### Format()
